---
### **(2) Read the telemetry files**
---

```python
PATH = "../iperf_client/"
files = os.listdir(PATH)

individual_telemetry = []
total_telemetry = []
cpu_utilization = []
for index, file in enumerate(files, 0):
  with open(PATH+file, "r") as f:
    data = json.load(f)
    intervals = data["intervals"]
    end = data["end"]

    metrics = []
    for interval in intervals:
      for streams in interval["streams"]:
        metrics.append({
          "start": streams["start"],
          "end": streams["end"],
          "bytes": streams["bytes"]
        })

    individual_telemetry.append({"run_id": index, "metrics": metrics}) 
    total_telemetry.append({"run_id": index, "sum_sent": end["sum_sent"], "sum_received": end["sum_received"]}) 
    cpu_utilization.append({"run_id": index, "cpu_utilization": end["cpu_utilization_percent"]})
```
